# Task 4 — Channel effectiveness, measured

**CLO1.** For *quantitative* data, how accurately a reader decodes a value depends on
the **visual channel** carrying it. Cleveland & McGill's ranking, best → worst:

```
position  >  length  >  angle  >  area  >  volume  >  colour
```

This is not taste — it comes from experiments measuring how accurately people judge
ratios. I reproduce it on a small scale: encode 30 value-pairs five ways, ask one reader
for the ratio each time, and rank the channels by error.

> **HOW TO RUN (easy version).** No pop-up questions. Step 2 shows all 30 trials as one
> grid; ask your reader for each, then type the 30 answers (0–100) into the list in the
> next cell and run it. The list refuses to continue until all 30 are filled.

````python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Wedge, Circle

plt.rcParams["figure.dpi"] = 110
ch = pd.read_csv("data/channel_trials.csv").drop(columns=["reader_estimate", "absolute_error"])
print(ch.shape)
print(ch["channel"].unique())
ch.head()
````

## 1. Encode a pair five ways

Five tiny chart functions, each showing one pair `(large, small)` through a single
channel. **No axis labels, no tick values, no numbers** — the reader must judge from the
channel alone.

**Attribute named per channel:** position (2-D position on a common scale), length (bar
extent from a shared zero), angle (wedge angle of a pie), area (circle area), colour
(lightness on a sequential ramp).

**Area honesty:** in `ax.scatter`, `s` **is area**. I set `s` proportional to the value
(not to its square), so area is proportional to value — the correct encoding, not the
classic area lie.

````python
def enc_position(ax, large, small):
    ax.scatter([0.4, 0.6], [large, small], s=140, color="#333333")
    ax.set_xlim(0, 1); ax.set_ylim(0, max(large, small) * 1.15)

def enc_length(ax, large, small):
    ax.bar([0.4, 0.6], [large, small], width=0.12, color="#333333")
    ax.set_xlim(0, 1); ax.set_ylim(0, max(large, small) * 1.15)   # common zero baseline

def enc_angle(ax, large, small):
    # two pies, each value as a wedge angle of a full circle (same max)
    top = max(large, small) * 1.15
    for cx, v in [(0.3, large), (0.7, small)]:
        ax.add_patch(Wedge((cx, 0.5), 0.18, 90 - 360 * v / top / 2, 90 + 360 * v / top / 2,
                           color="#333333"))
        ax.add_patch(Wedge((cx, 0.5), 0.18, 0, 360, fill=False, edgecolor="#cccccc"))
    ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.set_aspect("equal")

def enc_area(ax, large, small):
    # s is AREA in scatter; s proportional to value => area proportional to value
    ax.scatter([0.33, 0.67], [0.5, 0.5], s=[large * 22, small * 22], color="#333333")
    ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.set_aspect("equal")

def enc_colour(ax, large, small):
    top = max(large, small)
    # sequential lightness ramp (NOT a rainbow): darker = larger
    for cx, v in [(0.32, large), (0.68, small)]:
        grey = 0.85 - 0.75 * (v / top)
        ax.add_patch(plt.Rectangle((cx - 0.12, 0.35), 0.24, 0.3, color=(grey, grey, grey)))
    ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.set_aspect("equal")

ENCODERS = {"position": enc_position, "length": enc_length, "angle": enc_angle,
            "area": enc_area, "colour": enc_colour}

# preview: the same pair (80, 40) through all five channels
fig, axes = plt.subplots(1, 5, figsize=(15, 3.2))
for ax, (name, fn) in zip(axes, ENCODERS.items()):
    fn(ax, 80, 40)
    ax.set_xticks([]); ax.set_yticks([])
    ax.set_title(name)
fig.suptitle("The same 40:80 pair through five channels (reader sees no numbers)",
             fontsize=12)
plt.show()
````

## 2 & 3. Collect the reader's estimates, compute the error

> **HOW TO DO THIS (easy, no pop-ups).** The cell below shows **all 30 trials at once**
> as a labelled grid. For each one, ask your reader *"the smaller shape is what % of the
> larger?"* and note the number (0–100). Then, in the next cell, type those 30 numbers
> into the list — **one per trial, in order**. Run it. That's it. The trials are shuffled
> so the pattern can't be learned, and no true answers are shown to the reader.

````python
# Shuffle once, fixed, so the reader cannot learn the pattern. This SAME order is used
# by the answer list in the next cell.
show = ch.sample(frac=1, random_state=42).reset_index(drop=True)

fig, axes = plt.subplots(6, 5, figsize=(15, 17))
for k, (ax, (_, r)) in enumerate(zip(axes.ravel(), show.iterrows()), 1):
    ENCODERS[r["channel"]](ax, r["value_large"], r["value_small"])
    ax.set_xticks([]); ax.set_yticks([])
    ax.set_title(f"Trial {k}", fontsize=11)
fig.suptitle("For EACH trial ask: the smaller is what % of the larger? (0-100)",
             fontsize=14, fontweight="bold")
fig.tight_layout(rect=[0, 0, 1, 0.98])
plt.show()
````

## 4. Rank the channels by mean absolute error

One bar per channel, **sorted**, **zero baseline**, error shown. Grey for the pack, one
colour for the **best** channel (the single pop-out = the winner the title names).

````python
rank = res.groupby("channel")["absolute_error"].mean().sort_values()
print(rank.round(3))

fig, ax = plt.subplots(figsize=(8, 4.5))
cols = ["#bbbbbb"] * len(rank)
cols[0] = "#4477aa"   # best channel highlighted
ax.bar(rank.index, rank.values, color=cols)
for i, v in enumerate(rank.values):
    ax.annotate(f"{v:.2f}", (i, v), xytext=(0, 3), textcoords="offset points",
                ha="center", fontsize=9)
ax.set_title("Position is decoded most accurately; colour worst — as predicted",
             loc="left", fontweight="bold")
ax.set_ylabel("Mean absolute error (ratio)  —  one reader")
ax.set_ylim(0, None)
for s in ("top", "right"):
    ax.spines[s].set_visible(False)
fig.savefig("channel_ranking.png", bbox_inches="tight")
plt.show()
````

**Where it agrees / disagrees.** My ranking is
`position < length < angle < area < colour` by error — the same order as Cleveland &
McGill's `position > length > angle > area > volume > colour` (I have no **volume**
channel, so it is absent). With a real reader the middle of the ranking (angle vs area)
is the part most likely to swap, because their errors are close and one noisy reader has
no power to separate them. Report whatever your real reader gives, including a swap.

## 5. Apply it (`iris.csv`): same finding, two channels

The finding: **petal length separates the three iris species**. I show it twice — once
on a **high-ranking** channel (**position**: a strip of petal length by species) and
once on a **low-ranking** one (**colour**: species mean petal length as a lightness
ramp). Same data, same message.

````python
iris = pd.read_csv("data/iris.csv")
sp = ["setosa", "versicolor", "virginica"]
means = iris.groupby("species")["petal_length"].mean().reindex(sp)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))

# HIGH-ranking: position on a common scale
for i, s in enumerate(sp):
    vals = iris[iris["species"] == s]["petal_length"]
    jitter = np.random.default_rng(1).normal(i, 0.06, len(vals))
    axes[0].scatter(jitter, vals, color="#bbbbbb", s=18, edgecolors="none")
    axes[0].scatter(i, means[s], color="#ee6677", s=80, zorder=3)
axes[0].set_xticks(range(3)); axes[0].set_xticklabels(sp)
axes[0].set_ylabel("Petal length (cm)"); axes[0].set_ylim(0, None)
axes[0].set_title("POSITION (high rank): species separate cleanly", loc="left",
                  fontweight="bold")
for s in ("top", "right"):
    axes[0].spines[s].set_visible(False)

# LOW-ranking: colour lightness only
top = means.max()
for i, s in enumerate(sp):
    grey = 0.9 - 0.8 * (means[s] / top)
    axes[1].add_patch(plt.Rectangle((i, 0), 0.8, 1, color=(grey, grey, grey)))
    axes[1].text(i + 0.4, -0.12, s, ha="center")
axes[1].set_xlim(-0.2, 3); axes[1].set_ylim(-0.2, 1.05)
axes[1].set_xticks([]); axes[1].set_yticks([])
axes[1].set_title("COLOUR (low rank): how much longer? hard to say", loc="left",
                  fontweight="bold")
for s in ("top", "right", "left", "bottom"):
    axes[1].spines[s].set_visible(False)
fig.savefig("iris_two_channels.png", bbox_inches="tight")
plt.show()
````

**Which I would publish, and why.** The **position** version. On position the reader
reads each species' petal length off a common scale and judges the gaps directly: setosa
is obviously short, virginica obviously long, and you can quantify the difference. On the
**colour** (lightness) version you can tell the three shades apart but **cannot decode by
how much** they differ — lightness has poor perceptual linearity and no scale. Same data,
same message, but position lets the reader decode an accurate ratio and colour does not.
The choice is about **decoding accuracy**, not which looks nicer.

## The question — what 30 points from one reader can and cannot support

**Can support:** a *within-reader* demonstration that the *direction* of the
Cleveland–McGill ranking reproduces — position clearly beats colour for this reader.
**Cannot support:** any claim about the population, the exact size of each gap, or fine
distinctions between adjacent channels (angle vs area). Six ratios × one reader = no
replication and no statistical power; a single swap is noise.

**Why follow the published ranking anyway.** It is built on large, replicated,
controlled experiments. My tiny sample is consistent with it but could never overturn
it — so when my 30 points disagree on a detail, the rational move is to trust the large
study, not my noise. A small experiment is for *feeling* the effect, not for *re-deciding*
the ranking.
